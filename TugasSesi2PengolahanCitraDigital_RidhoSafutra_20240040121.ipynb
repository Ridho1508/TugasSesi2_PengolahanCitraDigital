```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(10,6))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Citra Asli (RGB)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_gray, cmap="gray")
plt.title("Citra Grayscale")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

print("Shape", gray.shape)

print("Nilai piksel: ")
print(gray[100, 200])
```

```python
#kuantisasi citra

def quartize_image(img, bits):
  levels = 2 ** bits
  step = 256 // levels
  quartized = (img // step) * step

  return quartized.astype(np.uint8)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

q8 = quartize_image(gray, 8)
q4 = quartize_image(gray, 4)
q2 = quartize_image(gray, 2)
q1 = quartize_image(gray, 1)

plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.imshow(q8, cmap="gray")
plt.title("8 Bit")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(q4, cmap="gray")
plt.title("4 Bit")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(q2, cmap="gray")
plt.title("2 Bit")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(q1, cmap="gray")
plt.title("1 Bit")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Chanel R G B pada citra

R = img_rgb[:,:,0]
G = img_rgb[:,:,1]
B = img_rgb[:,:,2]

#menampilkan hasil
plt.figure(figsize=(15,4))

plt.subplot(1,3,1)
plt.imshow(R, cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(G, cmap="gray")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(B, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Grayscale to Binary dg Python
threshold = 128

binary = np.where(gray >= threshold, 255, 0).astype(np.uint8)

plt.imshow(binary, cmap="gray")
plt.title(f"Threshold = {threshold}")
plt.axis("off")
plt.show()


threshold = 80

binary = np.where(gray >= threshold, 255, 0).astype(np.uint8)

plt.imshow(binary, cmap="gray")
plt.title(f"Threshold = {threshold}")
plt.axis("off")
plt.show()


threshold = 180

binary = np.where(gray >= threshold, 255, 0).astype(np.uint8)

plt.imshow(binary, cmap="gray")
plt.title(f"Threshold = {threshold}")
plt.axis("off")
plt.show()

threshold = 39

binary = np.where(gray >= threshold, 255, 0).astype(np.uint8)

plt.imshow(binary, cmap="gray")
plt.title(f"Threshold = {threshold}")
plt.axis("off")
plt.show()
```
